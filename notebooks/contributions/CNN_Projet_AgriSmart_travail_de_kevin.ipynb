# Task
Préparer les données pour le CNN en utilisant `ImageDataGenerator` pour le 'Corn' dataset, incluant le redimensionnement, la normalisation, l'augmentation de données pour l'entraînement, et la création de générateurs de données pour les ensembles d'entraînement et de validation.

## Préparer les données pour le CNN

### Subtask:
Prétraiter les images (redimensionnement, normalisation, augmentation de données si nécessaire) et les charger en utilisant `ImageDataGenerator` pour les ensembles d'entraînement et de validation du 'Corn' dataset.


**Reasoning**:
First, I'll import the `ImageDataGenerator` class, define the target image dimensions and batch size, and then initialize two `ImageDataGenerator` instances: one for training with data augmentation and rescaling, and another for validation with only rescaling, as per instructions 1-4.



```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Define target dimensions and batch size
IMG_HEIGHT = 224
IMG_WIDTH = 224
BATCH_SIZE = 32

# Create an ImageDataGenerator for the training set with data augmentation and rescaling
train_image_generator = ImageDataGenerator(
    rescale=1./255,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Create an ImageDataGenerator for the validation set with only rescaling
validation_image_generator = ImageDataGenerator(rescale=1./255)
```

**Reasoning**:
Next, I will use the `flow_from_directory()` method on both the training and validation `ImageDataGenerator` instances to load the images from their respective directories, specifying the target size, batch size, class mode, and the `corn_directories` as per instructions 5 and 6. Finally, I will extract the number of classes from the `train_generator` as per instruction 7.



```python
train_generator = train_image_generator.flow_from_directory(
    train_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=corn_directories
)

validation_generator = validation_image_generator.flow_from_directory(
    val_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=corn_directories
)

num_classes = train_generator.num_classes

print(f"Number of training samples: {train_generator.samples}")
print(f"Number of validation samples: {validation_generator.samples}")
print(f"Number of classes: {num_classes}")
```

## Définir l'architecture du baseline CNN

### Subtask:
Construire un modèle de réseau de neurones convolutif simple (CNN) en utilisant une bibliothèque comme TensorFlow/Keras, incluant des couches convolutives, de pooling, et des couches denses pour la classification des types de 'Corn' (maïs).


**Reasoning**:
I will import the necessary Keras layers and `Sequential` model, then construct the CNN architecture as specified, including convolutional, pooling, flatten, and dense layers, and finally display the model summary.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Activation

# Initialize a Sequential model
model = Sequential()

# Add the first Convolutional layer
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(IMG_HEIGHT, IMG_WIDTH, 3)))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Add the second Convolutional layer
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Flatten the output before feeding to dense layers
model.add(Flatten())

# Add a Dense layer
model.add(Dense(128, activation='relu'))

# Add the output layer
model.add(Dense(num_classes, activation='softmax'))

# Print the model summary
model.summary()
```

**Reasoning**:
The next logical step after defining the model architecture is to compile the model by specifying the optimizer, loss function, and metrics. I will use the Adam optimizer and categorical crossentropy loss, along with accuracy as a metric, suitable for multi-class classification.



```python
from tensorflow.keras.optimizers import Adam

# Compile the model
model.compile(
    optimizer=Adam(),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

## Entraîner le CNN

### Subtask:
Compiler le modèle avec un optimiseur (par exemple, Adam) et une fonction de perte (par exemple, 'categorical_crossentropy' ou 'sparse_categorical_crossentropy'), puis entraîner le CNN sur les données d'entraînement.


**Reasoning**:
Now that the model is compiled, I will train the CNN using the `fit()` method with the `train_generator` and `validation_generator` for a specified number of epochs, and store the training history.



```python
EPOCHS = 15

history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=validation_generator
)

print("Model training complete!")
```

## Évaluer le CNN et afficher les premiers résultats

### Subtask:
Évaluer la performance du modèle entraîné sur l'ensemble de validation et afficher des métriques clés telles que la précision et la perte. Cela constituera les 'Premiers résultats du CNN'.


**Reasoning**:
I will evaluate the trained model using the `evaluate()` method with the `validation_generator` to get the loss and accuracy on the validation set, then print these metrics as requested.



```python
eval_results = model.evaluate(validation_generator)

print(f"Validation Loss: {eval_results[0]:.4f}")
print(f"Validation Accuracy: {eval_results[1]:.4f}")
```

## Afficher les courbes d'entraînement du CNN

### Subtask:
Visualiser les courbes de précision et de perte de l'entraînement et de la validation au fil des époques pour comprendre le comportement du modèle et détecter le surapprentissage ou le sous-apprentissage.


**Reasoning**:
I will import `matplotlib.pyplot`, then plot the training and validation accuracy and loss over epochs using subplots, labels, titles, and legends as per the instructions to visualize the model's performance.



```python
import matplotlib.pyplot as plt

# Get training and validation metrics from history object
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(EPOCHS)

# Create a figure with two subplots
plt.figure(figsize=(12, 6))

# Plot training and validation accuracy
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(loc='lower right')

# Plot training and validation loss
plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(loc='upper right')

# Ensure proper spacing between subplots
plt.tight_layout()

# Display the plots
plt.show()
```

## Sauvegarder le modèle CNN

### Subtask:
